```python
# which year generated the most profit.
yearly_profit = df.groupby("Year")["Profit"].sum().reset_index()

yearly_profit["YoY Growth %"] = (
    yearly_profit["Profit"].pct_change() * 100
).round(2)

print(yearly_profit)
# 2025 showed stronger year-over-year growth in both sales and profit compared with 2024. Profit growth (9.00%) slightly exceeded sales growth (8.27%).
```

```python
# investigate where the sales come from ,Now we'll break the business down by Category.
category_analysis = df.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum"),
    Orders=("Order ID", "nunique")
).reset_index()

print(category_analysis)
```

Don't interpret:
"Highest quantity = highest sales."

Our data demonstrates why that's not necessarily true:
Office Supplies
Quantity = 10,625
Sales    = ₹5.53M

Technology
Quantity = 15,812
Sales    = ₹219.17M

Sales depend on both quantity and price, along with discount.

```python
# Calculate category profit margin
# How efficiently does each category convert sales into profit?
category_analysis["Profit Margin %"] = (
    category_analysis["Total_Profit"] /
    category_analysis["Total_Sales"]
) * 100

category_analysis["Profit Margin %"] = (
    category_analysis["Profit Margin %"].round(2)
)

print(category_analysis)
```

- Technology generates the largest sales and total profit in the dataset.
- Office Supplies has the highest profit margin at 16.85%, despite having the lowest total sales.
- Furniture has the lowest profit margin at 8.97%.
- This shows why we should look at both total profit and profit margin rather than using sales alone to judge category performance.
That's a good portfolio-level insight because you're distinguishing scale from profitability.

```python
# Visualize Category Performance
plt.figure(figsize=(8, 5))

plt.bar(
    category_analysis["Category"],
    category_analysis["Total_Sales"]
)

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.ticklabel_format(style="plain", axis="y")

plt.tight_layout()
plt.show()
```

```python
# profit chart
plt.figure(figsize=(8, 5))

plt.bar(
    category_analysis["Category"],
    category_analysis["Total_Profit"]
)

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")

plt.ticklabel_format(style="plain", axis="y")

plt.tight_layout()
plt.show()
```

Why two charts instead of one?
We're intentionally separating:
Sales → business scale
from
Profit → financial outcome

```python
# Shipping Analysis
# whether shipping time varies across shipping modes.
# calculate the average shipping days for each Ship Mode.
shipping_analysis = df.groupby("Ship Mode").agg(
    Average_Shipping_Days=("Shipping Days", "mean"),
    Orders=("Order ID", "nunique"),
    Total_Sales=("Sales", "sum")
).reset_index()

shipping_analysis["Average_Shipping_Days"] = (
    shipping_analysis["Average_Shipping_Days"].round(2)
)

print(shipping_analysis)
```

- Standard Class has the most orders: 5,505 and the highest sales: ₹158.03M.
- First Class has an average shipping time of 4.07 days.
- Same Day has 4.05 days in this dataset.
- Second Class has the lowest average: 4.00 days.
- The Unknown category contains only 7 orders, because we filled the original missing Ship Mode values with "Unknown".
* One important observation: the shipping times are very close to each other, so there isn't a large difference in average shipping duration among the known shipping modes.

```python
# Visualize Shipping Performance
plt.figure(figsize=(8, 5))

sns.barplot(
    data=shipping_analysis[shipping_analysis["Ship Mode"] != "Unknown"],
    x="Ship Mode",
    y="Average_Shipping_Days"
)

plt.title("Average Shipping Days by Ship Mode")
plt.xlabel("Ship Mode")
plt.ylabel("Average Shipping Days")
plt.xticks(rotation=15)
plt.show()
```

Loss-Making Analysis

```python
# Count loss-making records
loss_count = (df["Profit"] < 0).sum()

print("Loss-making records:", loss_count)
```

```python
# Calculate the total loss
total_loss = df.loc[df["Profit"] < 0, "Profit"].sum()

print("Total Loss:", total_loss)
```

```python
# loss-making sub-categories
loss_subcategories = df[df["Profit"] < 0].groupby("Sub-Category").agg(
    Loss_Records=("Profit", "count"),
    Total_Loss=("Profit", "sum")
).reset_index()

loss_subcategories = loss_subcategories.sort_values(
    "Total_Loss"
)

print(loss_subcategories)
```

Loss-Making Products

```python
# products with the largest total losses
loss_products = df[df["Profit"] < 0].groupby(
    "Product Name"
).agg(
    Loss_Records=("Profit", "count"),
    Total_Loss=("Profit", "sum"),
    Total_Sales=("Sales", "sum")
).reset_index()

loss_products = loss_products.sort_values(
    "Total_Loss"
)

print(loss_products.head(10))
```

```python
# Visualize the Top 10 loss-making products
top_loss_products = loss_products.head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_loss_products,
    x="Total_Loss",
    y="Product Name"
)

plt.title("Top 10 Loss-Making Products")
plt.xlabel("Total Loss")
plt.ylabel("Product")
plt.show()
```
